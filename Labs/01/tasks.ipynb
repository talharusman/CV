```python
class GroceryManager:

    def __init__(self):
        self.items = []

    def add_item(self, item, quantity, price):
        grocery = {
            "name": item,
            "quantity": quantity,
            "price": price
        }

        self.items.append(grocery)
        print(item, "was added to the list.")

    def remove_item(self, item):
        for grocery in self.items:
            if grocery["name"].lower() == item.lower():
                self.items.remove(grocery)
                print(item, "was removed from the list.")
                return

        print("Sorry,", item, "is not in the list.")

    def view_list(self):
        if len(self.items) == 0:
            print("The grocery list is empty.")
        else:
            print("\nGrocery List:")

            for grocery in self.items:
                print(
                    grocery["name"],
                    "- Quantity:",
                    grocery["quantity"],
                    "- Price: $",
                    grocery["price"]
                )

    def calculate_total(self):
        total = 0

        for grocery in self.items:
            total = total + (grocery["quantity"] * grocery["price"])

        return total


# Creating the grocery manager
grocery_list = GroceryManager()

# Adding some items
grocery_list.add_item("Apples", 2, 3.50)
grocery_list.add_item("Milk", 1, 2.50)
grocery_list.add_item("Bread", 2, 2.00)

# Display the list
grocery_list.view_list()

# Display the total
print("\nTotal cost: $", grocery_list.calculate_total())

# Remove an item
grocery_list.remove_item("Milk")

# Try removing an item that does not exist
grocery_list.remove_item("Eggs")

# Display the updated list
grocery_list.view_list()

print("\nFinal total: $", grocery_list.calculate_total())
```

```python
# 1
# Nested dictionary: student ID -> {Name, Major, Grades}
students = {
    101: {
        'Name': 'Alice Johnson',
        'Major': 'Computer Science',
        'Grades': [88, 92, 79, 95, 84]
    },
    102: {
        'Name': 'Bob Smith',
        'Major': 'Mathematics',
        'Grades': [70, 75, 68, 80, 72]
    },
    103: {
        'Name': 'Carol White',
        'Major': 'Computer Science',
        'Grades': [95, 98, 92, 97, 99]
    },
    104: {
        'Name': 'David Brown',
        'Major': 'Physics',
        'Grades': [60, 65, 70, 58, 63]
    },
    105: {
        'Name': 'Eva Green',
        'Major': 'Mathematics',
        'Grades': [85, 88, 90, 83, 87]
    }
}

print("Student Records Created Successfully!")
print(f"Total students: {len(students)}")

# 2
def get_highest_average_student(students):

    if not students:
        return "No students found."

    best_student = None
    highest_avg = -1

    for student_id, info in students.items():
        grades = info['Grades']
        avg = sum(grades) / len(grades)
        print(f"  Student ID {student_id} | {info['Name']} | Avg: {avg:.2f}")

        if avg > highest_avg:
            highest_avg = avg
            best_student = info['Name']

    return best_student, highest_avg


print("=== Average Grades ===")
top_student, top_avg = get_highest_average_student(students)
print(f"\n🏆 Student with Highest Average Grade: {top_student} ({top_avg:.2f})")


# 3

def search_by_major(students, major):

    print(f"\n=== Students enrolled in '{major}' ===")
    found = False

    for student_id, info in students.items():
        if info['Major'].lower() == major.lower():
            grades = info['Grades']
            avg = sum(grades) / len(grades)
            print(f"  ID: {student_id} | Name: {info['Name']} | Avg Grade: {avg:.2f}")
            found = True

    if not found:
        print(f"  No students found in '{major}'.")


# Test the search function
search_by_major(students, 'Computer Science')
search_by_major(students, 'Mathematics')
search_by_major(students, 'Physics')
search_by_major(students, 'Biology')   # Not found case

```
